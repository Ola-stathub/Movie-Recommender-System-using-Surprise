```python
all_movies = df['movieId'].unique()
```

```python
user_id = 138301
```

```python
rated_movies = df[df['userId'] == user_id]['movieId']
```

```python
rated_movies
```

```python
unrated_movies = []
```

```python
for movie in all_movies:
    if movie not in rated_movies.values:
        unrated_movies.append(movie)
```

```python
predictions = []
for movie_id in unrated_movies:
    pred = model.predict(user_id, movie_id)
    predictions.append(pred)
    
```

```python
predictions
```

```python
recc = sorted(predictions, key = lambda x: x.est, reverse = True)
```

```python
top_n = recc[:20]
```

```python
for pred in top_n:
    title = df[df['movieId'] == pred.iid]['title'].values[0]
    print(f'Reccommended: {title}: predicted rating: {pred.est:.2f}')
```
